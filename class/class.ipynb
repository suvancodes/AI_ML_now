### Check if a Number is an Armstrong Number

```python
a= int(input("Enter a number: "))
digit = len(str(a))
temp = a
sum = 0
while(temp!=0):
    num = temp%10
    sum = sum+num**digit
    temp = temp//10

if(a == sum):
    print("Armstrong number")
    
else:
    print("Not an Armstrong number")

```

### WAP to find factriol of a number

```python
# WAP to find factriol of a number using recursion 
a = int(input("Enter a number: "))
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print(f"The factorial of {a} is {factorial(a)}")
```

### WAP to print fibonacci series up to n terms

```python
## WAP to print fibonacci series up to n terms using recursion
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)


n = int(input("Enter number of terms: "))

print("Fibonacci Series:")

for i in range(n):
    print(fibonacci(i), end=" ")
```

### WAP to check the number is prime or not using recursion

```python
## WAP to check the number is prime or not using recursion
a = int(input("Enter a number: "))
def is_prime(n, i=2):
    if n <= 1:
        return False
    if i >= n:
        return True
    if n % i == 0:
        return False
    return is_prime(n, i + 1)

print(f"The number {a} is {'prime' if is_prime(a) else 'not prime'}")
```

### write a program to find a element in list using binary search 

```python

def binary_search(list, element, st, end):
    if st > end:
        return -1
    mid = (st + end) // 2
    if list[mid] == element:
        return mid
    elif list[mid] < element:
        return binary_search(list, element, mid + 1, end)
    else:
        return binary_search(list, element, st, mid - 1)

list = [1, 2, 3, 4, 5, 6, 7, 8, 9]
element = int(input("Enter the element to search: "))
result = binary_search(list, element, 0, len(list) - 1)

if result != -1:
    print(f"Element found at index: {result}")
else:
    print("Element not found in the list")
```

# n-queen problem

```python
def is_safe(board, row, col, n):

    # Check left side of row
    for i in range(col):
        if board[row][i] == 1:
            return False

    # Check upper-left diagonal
    i, j = row, col
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Check lower-left diagonal
    i, j = row, col
    while i < n and j >= 0:
        if board[i][j] == 1:
            return False
        i += 1
        j -= 1

    return True


def solve_n_queens(board, col, n):

    # Base Case
    if col >= n:

        print("\nSolution:\n")

        for row in board:
            for cell in row:
                if cell == 1:
                    print("👑", end=" ")
                else:
                    print("❌", end=" ")
            print()

        return True

    # Try placing queen in every row
    for i in range(n):

        if is_safe(board, i, col, n):

            # Place Queen
            board[i][col] = 1

            # Recursive Call (Backtracking)
            if solve_n_queens(board, col + 1, n):
                return True

            # Backtrack
            board[i][col] = 0

    return False


n = int(input("Enter number of queens: "))

board = [[0 for _ in range(n)] for _ in range(n)]

if not solve_n_queens(board, 0, n):
    print("No Solution Exists")
```
