### Step 1: Load Dataset

````python
import pandas as pd
import numpy as np
import pandas as pd

df = pd.read_csv(
    "SMSSpamCollection.txt",
    sep="\t",
    names=["label","message"]
)

df.head()
````

### Step 2: Text Preprocessing

````python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
````

### Step 3: Create Preprocessing Function

````python
def preprocess_text(text):
    words = word_tokenize(text)
    words = [word.lower() for word in words if word.isalpha()]
    words = [word for word in words if word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]
    return ' '.join(words)
````

````python
df["clean_message"] = df["message"].apply(preprocess_text)

df.head()
````

### Why use POS Tagging in a project?

POS tagging identifies the grammatical role of each word:

```text
Win      → Verb (VB)
free     → Adjective (JJ)
money    → Noun (NN)
```

### Importance in Spam Detection

* Helps understand the meaning of text.
* Can be used to extract only important words (nouns, verbs, adjectives).
* Improves feature engineering.
* Useful for advanced NLP tasks like NER, sentiment analysis, and information extraction.

### Example

Message:

```text
"Win free money now"
```

POS Tags:

```python
[
 ('Win', 'VB'),
 ('free', 'JJ'),
 ('money', 'NN'),
 ('now', 'RB')
]
```

Here you learn:

* **Win** → action (Verb)
* **Free** → description (Adjective)
* **Money** → object (Noun)

Spam messages often contain many verbs and promotional nouns like:

```text
win, claim, receive, call, prize, cash, offer
```

So POS tags can become additional features for your ML model.

For a beginner spam classifier, POS tagging is mostly **for learning and analysis**. For an advanced NLP project, it can be used as an actual feature to improve performance. 🚀


````python
sample = df["message"][0]

tokens = word_tokenize(sample)

pos_tags = nltk.pos_tag(tokens)

print(pos_tags)
````

### Step 6: Named Entity Recognition


### Short Interview Answer

"NER helps identify important entities such as names, organizations, locations, phone numbers, and money amounts from text. In spam detection, these entities can be used as additional features to improve classification performance."

````python
from nltk import ne_chunk

tokens = word_tokenize(df["message"][0])

pos_tags = nltk.pos_tag(tokens)

entities = ne_chunk(pos_tags)

print(entities)
````
