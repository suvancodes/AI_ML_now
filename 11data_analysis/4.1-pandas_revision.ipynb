## 1. Pandas Fundamentals

### What is Pandas?

**Pandas** is a Python library used for:

- Data manipulation
- Data cleaning
- Data analysis
- Reading/writing structured data
- Aggregation and transformation

```python
import pandas as pd
```

### Series

A **Series** is a one-dimensional labeled data structure.

```python
data = [1, 2, 3, 4, 5]
s = pd.Series(data)
```

With custom index:

```python
s = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
```

From dictionary:

```python
data = {'a': 1, 'b': 2, 'c': 3}
s = pd.Series(data)
```

### Interview Question

**Q: Series vs DataFrame?**

| Series | DataFrame |
|---|---|
| 1-dimensional | 2-dimensional |
| Single column-like | Multiple columns |
| Has index | Has row + column labels |

# 7. Missing Values ⭐

This is **very important for Data Analyst interviews**.

### Detect

```python
df.isnull()
```

### Fill with constant

```python
df.fillna(0)
```

### Fill with mean

Your notebook uses:

```python
df['Price'] = df['Price'].fillna(
    df['Price'].mean()
)
```

Conceptually:

```text
Missing numerical value
        ↓
Calculate mean
        ↓
Replace NaN with mean
```

### Interview Question

**Q: When would you use mean, median, or mode?**

- **Mean** → numerical data without strong outliers
- **Median** → numerical data with outliers/skew
- **Mode** → categorical data

⚠️ Your notebook specifically demonstrates **mean imputation**. The broader rule above is interview knowledge.

One of the most important Pandas concepts for DA interviews.

Example:

```python
df.groupby('Category')['Price'].mean()
```

Meaning:

```text
Group data by Category
        ↓
Select Price
        ↓
Calculate mean
```

## Multiple Aggregations

Your notebook:

```python
df.groupby('Category')['Price'].agg(
    ['sum', 'mean', 'count']
)
```

This calculates:

- `sum` → total
- `mean` → average
- `count` → number of records

### Interview Question

**Q: Why use `groupby()`?**

To perform calculations separately for different groups/categories.

For example:

> Find average sales by product category.

```python
df.groupby('Category')['Sales'].mean()
```

# 12. `sum(axis=1)` ⭐

Your notebook calculates total marks:

```python
df['total'] = df[
    ['Maths', 'Physics', 'Chemistry']
].sum(axis=1)
```

### Why `axis=1`?

It means **calculate across columns for each row**.

Example:

```text
Maths Physics Chemistry
 80     90       70
```

Result:

```text
240
```

### Remember

```text
axis=0 → down the rows / column-wise
axis=1 → across columns / row-wise
```

# 15. Merge / Join ⭐⭐⭐

Your notebook uses:

```python
pd.merge(
    df1,
    df2,
    left_on='key1',
    right_on='key2',
    how='inner'
)
```

The important parameter is:

```python
how=
```

---

## Inner Join

```python
pd.merge(df1, df2, how='inner')
```

Returns only matching records.

```text
A   B   C
    ↓
A   B   F
```

Result:

```text
A
B
```

---

## Left Join

```python
pd.merge(df1, df2, how='left')
```

Keeps **all rows from the left DataFrame**.

---

## Right Join

```python
pd.merge(df1, df2, how='right')
```

Keeps **all rows from the right DataFrame**.

---

## Outer Join

```python
pd.merge(df1, df2, how='outer')
```

Keeps **all rows from both DataFrames**.

### ⭐ Memorize this

| Join | Keeps |
|---|---|
| Inner | Matching rows |
| Left | Everything from left |
| Right | Everything from right |
| Outer | Everything from both |
