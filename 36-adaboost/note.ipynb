# AdaBoost — Mathematical Intuition & Revision Notes

## 1. What is AdaBoost?

**AdaBoost = Adaptive Boosting**

AdaBoost is an **ensemble learning technique** that combines multiple **weak learners** sequentially to create a **strong learner**.

### Core Idea

> AdaBoost gives more importance to the data points that previous weak learners classified incorrectly.

```text
Dataset
   ↓
Weak Learner 1
   ↓
Identify mistakes
   ↓
Increase weight of mistakes
   ↓
Weak Learner 2
   ↓
Identify mistakes
   ↓
Increase weight of mistakes
   ↓
...
   ↓
Strong Learner
```

---

# 2. Ensemble Learning

Ensemble learning combines multiple models to create a stronger model.

There are two important approaches:

| Technique    | Main Idea                  |
| ------------ | -------------------------- |
| **Bagging**  | Train models independently |
| **Boosting** | Train models sequentially  |

### Bagging

Example:

* Random Forest Classifier
* Random Forest Regressor

Models are trained independently and their predictions are combined.

### Boosting

Examples:

* AdaBoost
* Gradient Boosting
* XGBoost

Models are trained **sequentially**, where later models try to improve the mistakes of earlier models.

---

# 3. Weak Learner

A **weak learner** is a model that performs only slightly better than random guessing.

In AdaBoost, a common weak learner is a:

> **Decision Tree Stump**

### Decision Tree Stump

A decision stump is a very shallow decision tree, typically:

[
\text{Depth} = 1
]

Example:

```text
             Salary ≤ 50K?
              /        \
            Yes         No
            ↓            ↓
          Class A      Class B
```

A single stump usually has:

* High bias
* Low variance
* Relatively poor predictive power

But multiple weak learners can be combined to create a strong learner.

---

# 4. AdaBoost Mathematical Intuition

Suppose we have decision-tree stumps:

[
M_1,M_2,M_3,\ldots,M_n
]

AdaBoost combines their predictions using learner weights:

[
F(x)
====

\alpha_1M_1(x)
+
\alpha_2M_2(x)
+
\alpha_3M_3(x)
+\cdots+
\alpha_nM_n(x)
]

where:

* (M_i) = weak learner
* (\alpha_i) = importance/weight of that learner

### Important Intuition

```text
Better weak learner
        ↓
Higher α

Worse weak learner
        ↓
Lower α
```

So a more accurate stump gets more influence on the final prediction.

---

# 5. AdaBoost Algorithm

The complete AdaBoost training process can be remembered in these steps:

```text
1. Initialize equal sample weights
          ↓
2. Train weak learner
          ↓
3. Calculate weighted error
          ↓
4. Calculate learner weight α
          ↓
5. Update sample weights
          ↓
6. Normalize weights
          ↓
7. Train next weak learner
          ↓
8. Repeat
          ↓
9. Combine all weak learners
```

---

# 6. Step 1 — Initialize Sample Weights

Suppose the training dataset contains (N) samples.

Initially, every sample receives equal importance.

[
\boxed{w_i=\frac{1}{N}}
]

For 7 samples:

[
w_i=\frac{1}{7}
]

Therefore:

```text
Sample 1 → 1/7
Sample 2 → 1/7
Sample 3 → 1/7
Sample 4 → 1/7
Sample 5 → 1/7
Sample 6 → 1/7
Sample 7 → 1/7
```

---

# 7. Step 2 — Train the First Decision Stump

Train a weak learner using the current sample weights.

The example in the notes uses features such as:

* Salary
* Credit
* Approval

The decision stump may choose a split such as:

```text
Salary ≤ 50K?
```

or:

```text
Credit = G?
```

The decision tree uses impurity measures such as:

* Entropy
* Gini Impurity

---

# 8. Entropy

The entropy formula shown in the notes is:

[
H(S)
====

## -P_1\log_2(P_1)

P_2\log_2(P_2)
]

More generally:

[
\boxed{
H(S)
====

-\sum_i P_i\log_2(P_i)
}
]

### Intuition

```text
High Entropy
     ↓
More mixed classes

Low Entropy
     ↓
More pure classes
```

A decision tree tries to find splits that reduce impurity.

---

# 9. Step 3 — Calculate Total Error

After training the stump, check which samples were classified incorrectly.

The weighted total error is:

[
\boxed{
TE
==

\sum_{\text{incorrect samples}}w_i
}
]

For example, if there are 7 samples and one sample is classified incorrectly:

[
TE=\frac{1}{7}
]

Therefore:

[
TE\approx0.143
]

---

# 10. Step 4 — Calculate Performance of the Stump

The weight/performance of the weak learner is:

[
\boxed{
\alpha
======

\frac{1}{2}
\ln
\left(
\frac{1-TE}{TE}
\right)
}
]

where:

* (TE) = Total Error
* (\alpha) = Weight of the weak learner

### Example

If:

[
TE=\frac{1}{7}
]

then:

[
\alpha
======

\frac{1}{2}
\ln
\left(
\frac{1-\frac17}{\frac17}
\right)
]

# [

\frac{1}{2}\ln(6)
]

[
\boxed{\alpha\approx0.896}
]

### Important Relationship

[
TE\downarrow
\quad\Rightarrow\quad
\alpha\uparrow
]

Therefore:

> **A more accurate weak learner receives a larger weight.**

---

# 11. Step 5 — Update Sample Weights

This is the **most important idea in AdaBoost**.

After the weak learner makes predictions:

### Correctly Classified Samples

Their weights are **decreased**.

[
\boxed{
w_i'
====

w_i e^{-\alpha}
}
]

### Incorrectly Classified Samples

Their weights are **increased**.

[
\boxed{
w_i'
====

w_i e^{+\alpha}
}
]

---

## Example

Initial weight:

[
w_i=\frac17
]

For a correctly classified sample:

[
w_i'
====

\frac17e^{-0.896}
]

[
\approx0.058
]

For an incorrectly classified sample:

[
w_i'
====

\frac17e^{0.896}
]

[
\approx0.349
]

Therefore:

```text
Correctly classified
        ↓
Weight decreases
        ↓
≈ 0.058


Incorrectly classified
        ↓
Weight increases
        ↓
≈ 0.349
```

---

# 12. Why Increase the Weight of Incorrect Samples?

This is the **core intuition of AdaBoost**.

Suppose:

```text
Weak Learner 1
      ↓
Makes mistake on Sample X
```

AdaBoost increases the importance of Sample X.

Then:

```text
Sample X
   ↓
Weight ↑
   ↓
Next weak learner pays more attention
   ↓
Attempts to correct the previous mistake
```

Therefore:

> **Each new weak learner focuses more on the difficult examples.**

---

# 13. Step 6 — Normalize the Weights

After updating the sample weights, normalize them.

The formula is:

[
\boxed{
w_i^{new}
=========

\frac{w_i'}{\sum_j w_j'}
}
]

The purpose is to make the weights form a probability distribution:

[
\sum_i w_i^{new}=1
]

### Example

Suppose the updated weights are approximately:

```text
Correct samples   → 0.058
Incorrect sample  → 0.349
```

After normalization, the incorrectly classified sample gets a much larger probability.

In the example from the notes, it receives approximately:

[
0.50
]

while other samples receive approximately:

[
0.08
]

---

# 14. Probability / Roulette-Wheel Interpretation

After normalization, each sample gets a probability interval.

Example:

```text
Sample 1 → 0.00 – 0.08
Sample 2 → 0.08 – 0.16
Sample 3 → 0.16 – 0.24
Sample 4 → 0.24 – 0.32
Sample 5 → 0.32 – 0.40
Sample 6 → 0.40 – 0.70
Sample 7 → 0.70 – 0.78
```

A random number can then be generated to select samples.

### Important Intuition

```text
Higher sample weight
        ↓
Higher probability of selection
        ↓
Sample appears more often
        ↓
Next weak learner focuses on it
```

---

# 15. Step 7 — Train the Next Weak Learner

The new weighted dataset is given to the next decision stump.

```text
Weighted Dataset
       ↓
Decision Stump 2
       ↓
Calculate Error
       ↓
Calculate α₂
       ↓
Update Weights
       ↓
Normalize
       ↓
Next Stump
```

The notes calculate another stump performance:

[
\alpha_2\approx0.65
]

Now the ensemble becomes:

[
F(x)
====

\alpha_1M_1(x)
+
\alpha_2M_2(x)
]

where:

[
\alpha_1\approx0.896
]

[
\alpha_2\approx0.65
]

---

# 16. Repeat the Process

The same process is repeated for multiple weak learners.

```text
Stump 1
   ↓
Calculate Error
   ↓
Calculate α₁
   ↓
Update Weights
   ↓
Stump 2
   ↓
Calculate Error
   ↓
Calculate α₂
   ↓
Update Weights
   ↓
Stump 3
   ↓
...
   ↓
Final Strong Model
```

---

# 17. Final Prediction

Suppose several weak learners produce:

```text
DT1 → Yes
DT2 → No
DT3 → Yes
DT4 → No
```

with learner weights:

[
\alpha_1=0.896
]

[
\alpha_2=0.50
]

[
\alpha_3=0.24
]

[
\alpha_4=-0.30
]

The final score is calculated using the weighted predictions:

[
F(x)
====

\alpha_1M_1(x)
+
\alpha_2M_2(x)
+
\alpha_3M_3(x)
+
\alpha_4M_4(x)
]

From the example in the notes:

[
F(x)
====

0.896(Yes)
+
0.50(No)
+
0.24(Yes)
---------

0.30(No)
]

The resulting scores are approximately:

[
\text{Score(Yes)}=1.136
]

[
\text{Score(No)}=0.30
]

Since:

[
1.136>0.30
]

the final prediction is:

[
\boxed{\text{Yes}}
]

# 18. Complete AdaBoost Flow

```text
                Training Dataset
                       ↓
              Initialize Weights
                  wᵢ = 1/N
                       ↓
             Train Decision Stump
                       ↓
              Calculate Total Error
                       ↓
            Calculate Learner Weight
                       ↓
        α = 1/2 ln((1 - TE) / TE)
                       ↓
              Update Sample Weights
                       ↓
          Correct → Weight decreases
        Incorrect → Weight increases
                       ↓
                Normalize Weights
                       ↓
              Weighted Dataset
                       ↓
             Train Next Stump
                       ↓
                     Repeat
                       ↓
          Weighted Combination of
             All Weak Learners
                       ↓
               Strong Learner
```

# 20. AdaBoost vs Random Forest

| Feature           | Random Forest      | AdaBoost                       |
| ----------------- | ------------------ | ------------------------------ |
| Ensemble          | Bagging            | Boosting                       |
| Training          | Independent        | Sequential                     |
| Base Models       | Decision Trees     | Weak learners / Stumps         |
| Sample Weighting  | Not the main idea  | Core concept                   |
| Main Focus        | Reduce variance    | Correct mistakes               |
| Learners          | Mostly independent | Dependent on previous learners |
| Final Combination | Voting / Averaging | Weighted combination           |

---

# 21. Most Important Intuition 🧠

Remember this sequence:

```text
Equal Weights
      ↓
Train Weak Learner
      ↓
Find Mistakes
      ↓
Increase Weight of Mistakes
      ↓
Train Next Learner
      ↓
Focus More on Difficult Samples
      ↓
Repeat
      ↓
Weighted Combination
      ↓
Strong Learner
```

### One-line definition

> **AdaBoost sequentially combines weak learners and adaptively increases the importance of incorrectly classified samples so that later learners focus more on previous mistakes.**

# 22. One-Minute Revision 🚀

Before an exam/interview, remember these **7 points**:

1. **AdaBoost = Adaptive Boosting**
2. Uses **weak learners**, commonly decision stumps.
3. Start with **equal sample weights**.
4. Calculate the weak learner's **weighted error**.
5. Calculate:

[
\alpha=
\frac12\ln\left(\frac{1-TE}{TE}\right)
]

6. **Correct → weight decreases; Incorrect → weight increases.**
7. Combine all weak learners using their **learner weights**.

### The core idea:

> **AdaBoost keeps making the next weak learner focus on the mistakes made by previous weak learners.** 🎯
