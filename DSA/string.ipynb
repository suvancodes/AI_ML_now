
# 11. Longest Substring Without Repeating Characters ⭐

This is **very important for placements**.

### Problem

Find the length of the longest substring containing no repeated characters.

**Input**

```text
abcabcbb
```

**Output**

```text
3
```

Because:

```text
abc
```

is the longest.

Another:

```text
pwwkew
```

Output:

```text
3
```

because:

```text
wke
```

### Algorithm

Use:

**Sliding Window + HashSet/HashMap**

Example:

```text
a b c a b c b b
↑     ↑
left  right
```

When duplicate appears, move `left`.

**Time:** `O(n)`
**Space:** `O(k)`


# 15. Longest Palindromic Substring ⭐

### Problem

**Input**

```text
babad
```

**Output**

```text
bab
```

or:

```text
aba
```

### Algorithm 1 — Expand Around Center

Every palindrome has a center.

For:

```text
aba
```

center = `b`.

For:

```text
abba
```

center lies between `b` and `b`.

Expand outward.

**Time:** `O(n²)`
**Space:** `O(1)`

### Advanced algorithm

**Manacher's Algorithm**

Time:

`O(n)`

This is worth knowing conceptually, but don't prioritize it over sliding window/KMP for typical placements.

# 20. KMP Pattern Matching ⭐

This is one of the most important string algorithms.

### Problem

Same pattern-searching problem, but improve the worst-case complexity.

**Input**

```text
text = "ababcabcabababd"
pattern = "ababd"
```

**Output**

```text
10
```

### Main concept

Build:

```text
LPS array
```

LPS = **Longest Proper Prefix which is also Suffix**

Instead of starting the pattern from zero after a mismatch, KMP tells you where to continue.

**Time:** `O(n + m)`
**Space:** `O(m)`

### Priority

For placements:

> **Learn the idea + implementation of LPS/KMP.**

# 21. Rabin-Karp Pattern Matching

### Problem

Find occurrences of a pattern in a text using hashing.

**Input**

```text
text = "abcabcabc"
pattern = "abc"
```

**Output**

```text
[0, 3, 6]
```

### Main idea

Calculate a **rolling hash** for each window.

Instead of comparing every character every time:

```text
window → hash
```

Compare hashes first.

**Average Time:** `O(n + m)`
**Worst Time:** `O(n × m)` due to collisions
**Space:** `O(1)`

# 22. Group Anagrams ⭐

### Problem

Group words that are anagrams.

**Input**

```text
["eat","tea","tan","ate","nat","bat"]
```

**Output**

```text
[
 ["eat","tea","ate"],
 ["tan","nat"],
 ["bat"]
]
```

### Algorithm

Create a frequency representation:

```text
eat → a1,e1,t1
tea → a1,e1,t1
ate → a1,e1,t1
```

Use this representation as a HashMap key.

**Time:** `O(n × k)`

where:

* `n` = number of strings
* `k` = average string length

**Space:** `O(n × k)`

# 🔴 23. Longest Repeating Character Replacement ⭐

### Problem

You can replace at most `k` characters.

Find the longest substring that can be converted into the same character.

**Input**

```text
s = "AABABBA"
k = 1
```

**Output**

```text
4
```

For example:

```text
AABA
```

Change `B → A`:

```text
AAAA
```

### Main idea

Sliding window.

Maintain:

```text
maxFrequency
windowLength
```

Window is valid when:

```text
windowLength - maxFrequency <= k
```

**Time:** `O(n)`
**Space:** `O(1)` for uppercase English alphabet.

This is a **must-practice sliding-window problem**.

# 🟡 24. Generate All Permutations of a String

### Problem

Generate all possible permutations.

**Input**

```text
ABC
```

**Output**

```text
ABC
ACB
BAC
BCA
CAB
CBA
```

Number of permutations:

```text
3! = 6
```

### Algorithm

**Backtracking**

Choose a character → recurse → undo choice.

**Time:** `O(n × n!)`

Why?

There are:

```text
n!
```

permutations, and constructing/outputting each one costs `O(n)`.

**Space:** `O(n)` recursion stack, excluding output.

# 🔴 25. Word Break ⭐

### Problem

Determine whether a string can be segmented into dictionary words.

**Input**

```text
s = "leetcode"

dictionary = ["leet", "code"]
```

**Output**

```text
true
```

Because:

```text
leet + code
```

### Another example

```text
s = "catsandog"

dictionary = ["cats","dog","sand","and","cat"]
```

Output:

```text
false
```

### Algorithm

Dynamic Programming.

Define:

```text
dp[i] = whether first i characters can be formed
```

Transition:

```text
dp[i] = true
```

if there exists a previous position `j` where:

```text
dp[j] = true
```

and:

```text
s[j:i]
```

is in the dictionary.

**Time:** roughly `O(n²)` with efficient set lookup, depending on substring/hash implementation.

**Space:** `O(n)`.


# 2. Check Whether a String Is Palindrome

### Problem

Determine whether a string reads the same forward and backward.

**Input**

```text
madam
```

**Output**

```text
true
```

**Input**

```text
hello
```

**Output**

```text
false
```

### Algorithm

Two pointers:

```text
i = 0
j = n-1
```

Compare `s[i]` and `s[j]`.

**Time:** `O(n)`
**Space:** `O(1)`

````python
s = 'madam'
i = 0
j = len(s)-1
f = True
while(i<j):
    if(s[i]!=s[j]):
        f = False
    i+=1
    j-=1
f
````



# 3. Count Frequency of Every Character

### Problem

Count how many times each character appears.

**Input**

```text
programming
```

**Output**

```text
p → 1
r → 2
o → 1
g → 2
a → 1
m → 2
i → 1
n → 1
```

### Algorithm

Use:

* Array of size 26 for lowercase English letters
* HashMap for general characters

**Time:** `O(n)`
**Space:** `O(1)` if alphabet is fixed.

---

````python
s = 'programming'
d = {}
for i in range(len(s)):
    if s[i] not in d:
        d[s[i]] = 1
    elif s[i] in d:
        d[s[i]]+=1
d
````

# 4. First Non-Repeating Character

### Problem

Find the first character that appears exactly once.

**Input**

```text
swiss
```

**Output**

```text
w
```

Because:

```text
s → 3
w → 1
i → 1
```

`w` occurs first.

### Algorithm

**Pass 1:** frequency count
**Pass 2:** find first frequency = 1

**Time:** `O(n)`
**Space:** `O(1)` for fixed alphabet.


````python
s = 'swiss'
d={}
for i in range(len(s)):
    if s[i] not in d:
        d[s[i]] = 1
    elif s[i] in d:
        d[s[i]]+=1
for i in d:
    if d[i] == 1:
        print(i)
````
