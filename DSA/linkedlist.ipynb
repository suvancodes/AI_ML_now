
# 🔗 Linked List — Part 1: Theory

## 1. What is a Linked List?

A **linked list** is a linear data structure where elements are stored in separate memory locations and connected using pointers/references.

Think of it like a chain:

```text
[10 | •] → [20 | •] → [30 | •] → [40 | NULL]
```

Each box is called a **Node**.

A node contains two things:

```text
┌───────────────┐
│ Data │ Next   │
└───────────────┘
```

* **Data** → actual value
* **Next** → address/reference of the next node

---

# 2. Why do we need Linked Lists?

You already know arrays.

### Array

```text
[10][20][30][40][50]
```

Array elements are stored in **contiguous memory**.

If you want to insert something at the beginning:

```text
[10][20][30][40]
 ↓
[5][10][20][30][40]
```

Existing elements may need to be shifted.

### Linked List

```text
10 → 20 → 30 → 40
```

Insert `5` at beginning:

```text
5 → 10 → 20 → 30 → 40
```

No shifting is required.

That's one of the major advantages of linked lists.

---

# 3. Node

The most important concept.

In Java, a simple node looks like:

```java
class Node {
    int data;
    Node next;
}
```

Suppose:

```java
Node n1 = new Node();
n1.data = 10;
```

We have:

```text
n1
 ↓
┌─────────┐
│ 10 | ?  │
└─────────┘
```

Now:

```java
Node n2 = new Node();
n2.data = 20;

n1.next = n2;
```

Now:

```text
n1
 ↓
┌─────────┐      ┌─────────┐
│ 10 | ───────→  │ 20 | ?  │
└─────────┘      └─────────┘
                   ↑
                  n2
```

---

# 4. What is `head`?

`head` is a reference pointing to the **first node**.

Example:

```text
head
 ↓
10 → 20 → 30 → 40 → NULL
```

If:

```java
Node head;
```

and:

```java
head = n1;
```

then:

```text
head
 ↓
[10] → [20] → [30] → [40] → NULL
```

### Important

There is usually **no separate reference to every node**.

You start from `head` and follow:

```text
head
 ↓
node1 → node2 → node3 → node4 → NULL
```

---

# 5. What is `NULL`?

The last node doesn't point to another node.

Therefore:

```text
10 → 20 → 30 → NULL
```

In Java:

```java
next = null;
```

`null` means:

> "There is no next node."

---

# 6. How do we traverse a Linked List?

Traversal means visiting every node.

Suppose:

```text
head
 ↓
10 → 20 → 30 → 40 → NULL
```

We create a temporary reference:

```java
Node temp = head;
```

Then:

```java
while (temp != null) {
    System.out.println(temp.data);
    temp = temp.next;
}
```

### Understand this VERY carefully

Initially:

```text
temp
 ↓
10 → 20 → 30 → 40 → NULL
```

After:

```java
temp = temp.next;
```

it becomes:

```text
       temp
        ↓
10 → 20 → 30 → 40 → NULL
```

Then again:

```text
10 → 20    30 → 40 → NULL
          ↑
         temp
```

And so on.

The fundamental linked-list movement is:

```java
temp = temp.next;
```

**Memorize the concept, not the line.**

---

# 7. Creating a Linked List

Basic implementation:

```java
class Node {
    int data;
    Node next;

    Node(int data) {
        this.data = data;
        this.next = null;
    }
}
```

Then:

```java
Node head = new Node(10);

head.next = new Node(20);
head.next.next = new Node(30);
```

Result:

```text
head
 ↓
10 → 20 → 30 → NULL
```

---

# 8. Singly Linked List

The linked list we've discussed is called a **Singly Linked List**.

Each node points only **forward**.

```text
10 → 20 → 30 → 40 → NULL
```

You can move:

```text
10 → 20 → 30 → 40
```

but not directly:

```text
40 → 30
```

because node `40` doesn't know where `30` is.

---

# 9. Doubly Linked List

A doubly linked list has two references:

```text
┌───────────────┐
│ Prev | Data | Next │
└───────────────┘
```

Example:

```text
NULL ← 10 ⇄ 20 ⇄ 30 ⇄ 40 → NULL
```

You can move:

```text
10 → 20 → 30 → 40
```

and:

```text
40 → 30 → 20 → 10
```

because every node knows both its previous and next node.

---

# 10. Circular Linked List

In a normal linked list:

```text
10 → 20 → 30 → NULL
```

In a circular linked list:

```text
10 → 20 → 30
↑         ↓
└─────────┘
```

The last node points back to the first node.

```text
10 → 20 → 30
↑         |
└─────────┘
```

So there is **no `NULL` at the end**.

---

# 11. Types of Linked Lists

You should know these four:

| Type            | Structure                    |
| --------------- | ---------------------------- |
| Singly          | `10 → 20 → 30 → NULL`        |
| Doubly          | `NULL ← 10 ⇄ 20 ⇄ 30 → NULL` |
| Circular Singly | `10 → 20 → 30 → 10`          |
| Circular Doubly | `10 ⇄ 20 ⇄ 30 ⇄ 10`          |

For placements, **Singly Linked List is the first priority**.

---

# 12. Array vs Linked List

This is important for interviews.

| Feature             | Array         | Linked List                |
| ------------------- | ------------- | -------------------------- |
| Memory              | Contiguous    | Non-contiguous             |
| Size                | Usually fixed | Dynamic                    |
| Random access       | Fast          | Slow                       |
| Access `arr[i]`     | `O(1)`        | `O(n)`                     |
| Insert at beginning | `O(n)`        | `O(1)`*                    |
| Delete at beginning | `O(n)`        | `O(1)`*                    |
| Extra memory        | Low           | Pointer/reference required |

`*` Assuming you already have the `head` reference and are discussing the basic linked-list operation.

### The key difference:

Array:

```text
index → directly access
```

Linked List:

```text
head → next → next → next
```

So to reach the 100th node, you generally have to walk through the previous nodes.

---

# 13. Time Complexity You Must Know

For a basic singly linked list:

| Operation           | Complexity |
| ------------------- | ---------: |
| Access by position  |     `O(n)` |
| Search              |     `O(n)` |
| Insert at beginning |     `O(1)` |
| Delete at beginning |     `O(1)` |
| Insert at end*      |     `O(n)` |
| Delete at end       |     `O(n)` |

`*` Can be `O(1)` if the implementation maintains an appropriate tail reference.

Don't blindly memorize these. Understand **why**.

For example:

### Search for 30

```text
head
 ↓
10 → 20 → 30 → 40
```

You must check:

```text
10 ❌
20 ❌
30 ✅
```

Therefore:

```text
O(n)
```

# 14. Insertion — The Core Idea

Suppose:

```text
10 → 20 → 30
```

You want to insert `15` between `10` and `20`.

Before:

```text
10 → 20 → 30
```

After:

```text
10 → 15 → 20 → 30
```

Conceptually:

```text
newNode.next = 10.next
10.next = newNode
```

Result:

```text
10 → 15 → 20 → 30
```

### ⚠️ Very important

Order matters.

If you change:

```text
10.next
```

before saving the old connection, you can lose part of the list.

This pointer manipulation is one of the most important skills in linked lists.

---

# 15. Deletion — Core Idea

Suppose:

```text
10 → 20 → 30
```

Delete `20`.

We want:

```text
10 → 30
```

Conceptually:

```text
10.next = 20.next
```

So:

```text
10 → 30
```

The node `20` is no longer connected to the list.

---

# 16. The Three References You Should Understand

While solving linked-list problems, you'll frequently see:

```text
head
temp
prev
```

### `head`

Points to the first node.

```text
head
 ↓
10 → 20 → 30
```

### `temp`

Used to move through the list.

```text
10 → 20 → 30
     ↑
    temp
```

### `prev`

Used when you need to remember the previous node.

```text
10 → 20 → 30
 ↑    ↑
prev temp
```

This becomes **extremely important** in deletion and reversal problems.

---

# 17. The Mental Model You Need

Don't think:

> "Linked list is a bunch of code."

Think:

> **A linked list is nodes connected by references.**

For example:

```text
head
 ↓
┌──────┬──────┐
│ 10   │  ─────────┐
└──────┴──────┘    ↓
                ┌──────┬──────┐
                │ 20   │  ─────────┐
                └──────┴──────┘    ↓
                                ┌──────┬──────┐
                                │ 30   │ NULL │
                                └──────┴──────┘
```

If you can mentally manipulate this diagram, you'll be able to solve linked-list problems.

### Creating a Linked List 

````python
class node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
class LL:
    def __init__(self):
        self.head = None
    
    def add(self,data):
        newnode = node(data)
        if self.head ==None:
            self.head = newnode
            return
        temp = self.head
        while temp.next!=None:
            temp=temp.next
        temp.next = newnode
        
    def add_head(self,data):
        newnode = node(data)
        newnode.next = self.head
        self.head = newnode
        
    def add_idx(self,data,i):
        newnode = node(data)
        temp = self.head
        k = 1
        if i==1:
            newnode.next = self.head
            self.head = newnode
            return
        
        while k!=i-1:
            temp = temp.next
            k+=1
            
        newnode.next = temp.next
        temp.next = newnode
    
     
    def delete(self, data):
        if self.head is None:
            return
        if self.head.data == data:
            self.head = self.head.next
            return
        temp = self.head
        while temp.next is not None:
            if temp.next.data == data:
                temp.next = temp.next.next
                return
            temp = temp.next
    
    
     
    def delete_head(self):
        if self.head == None:
            print('Empty Linked List')
            return
        temp = self.head
        self.head = temp.next
        temp.next = None
        
    
    
        

    def display(self):
        temp = self.head
        while temp:
            print(temp.data,end='->')
            temp = temp.next
        print("none")
    
    
    
ll = LL()
ll.add(10)
ll.add_head(1)
ll.add(20)
ll.add(30)
ll.add_idx(12,2)
ll.add_idx(13,1)
ll.add_idx(121,4)
ll.delete_head()

ll.display()
````
