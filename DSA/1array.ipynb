## 18. Intersection of Two Arrays

```text
A = [1, 2, 2, 3, 4]
B = [2, 2, 4, 6]

Intersection:
[2, 2, 4]
```

You need to pay attention to whether the question asks for:

* unique intersection, or
* intersection including duplicates.

**Target TC:** `O(N + M)` with hashing
**Space:** `O(N)`
**Time:** **20 min**

````python
arr1 = [1,5,2,2,3,4]
arr2 = [2,4,4,2,6]
ans = []
d = {}
for i in range(len(arr1)):
    d[arr1[i]] = d.get(arr1[i], 0)+1

for i in range(len(arr2)):
    if arr2[i] in d:
        ans.append(arr2[i])
        if d[arr2[i]] == 1:
            d.pop(arr2[i])
        else:
            d[arr2[i]] -= 1
        
ans
````

## 19. Union of Two Arrays

```text
A = [1, 2, 3,4]
B = [2, 3, 4, 5]

Output:
[1, 2, 3, 4, 5]
```

Duplicates should appear only once.

**Target TC:** `O(N + M)` average
**Space:** `O(N + M)`
**Time:** **20 min**

````python
a = [1,2,3,4]
b= [2,3,4,5,6,7,8,9]
n = len(a)
m = len(b)
j = 0
k = 0
ans = []
while j<n and k <m:
    if a[j]<b[k]:
        ans.append(a[j])
        j+=1
        
    elif a[j]==b[k]:
        ans.append(a[j])
        j+=1
        k+=1
        
    else:
        ans.append(b[k])
        k+=1
        
while j<n:
    ans.append(a[j])
    j+=1
while k<m:
    ans.append(b[k])
    k+=1


        

ans
````

## 31. Three Sum

```text
[-1,0,1,2,-1,-4]
```

Find all unique triplets whose sum is `0`.

Output:

```text
[-1,-1,2]
[-1,0,1]
```

Pattern:

> Sorting + Two Pointers

**Target:** `O(N²)`
**Time:** **35–45 min**

````python
arr = [-1,0,1,2,-1,-4]
arr.sort()
result = []
for i in range(len(arr)-2):
    l = i+1
    r = len(arr)-1
    while l<r:
        t = arr[l]+arr[r]+arr[i]
        if t == 0:
            result.append([arr[i],arr[l],arr[r]])
            break
        elif t < 0:
            l = l+1
        else:
            r = r-1
result
        
        
````

## 34. Longest Subarray With Sum K

```text
arr = [10,5,2,7,1,9]
K = 15

Output:
4
```

One valid subarray:

```text
[5,2,7,1]
```

Pattern:

> Prefix Sum + HashMap

**Target:** `O(N)`
**Space:** `O(N)`
**Time:** **30–40 min**

````python
arr = [10,5,2,7,1,9]
k = 15
p_s = 0
m_l = 0
p_map={0:-1}
l=0
for i in range(len(arr)):
    num = arr[i]
    p_s+=num
    nn = p_s-k
    if nn in p_map:
        l = i - p_map[nn]
        
    m_l = max(l,m_l)
    if nn not in p_map:
        p_map[p_s] = i
m_l
````

## 37. Sort 0s, 1s and 2s


> **Dutch National Flag Algorithm**

**Target:** `O(N)`
**Space:** `O(1)`

🔥 Very good placement problem.

````python
arr = [2,0,2,1,1,0]
n0 = 0
n1 = 0
n2=0

for i in range(len(arr)):
    if arr[i] == 0:
        n0+=1
    elif arr[i]==1:
        n1+=1
    else:
        n2+=1
for i in range(len(arr)):
    if i<n0:
        arr[i] = 0
    elif i<n0+n1:
        arr[i] = 1
    elif i < n0+n1+n2:
        arr[i]=2
    
arr
````

## 39. Equilibrium Index

````python
arr = [-7,1,5,2,-4,3,0]
n = sum(arr)
sum = arr[0]
for i in range(1,len(arr)):
    div = (n - arr[i])/2
    if div == sum:
        print(i)
        break
    
    sum += arr[i]
````

## 42. Longest Consecutive Sequence

```text
[100,4,200,1,3,2]

Output:
4
```

Because:

```text
1,2,3,4
```

Don't sort if you're targeting the optimal solution.

**Target:** `O(N)` average
**Space:** `O(N)`
**Time:** **30–40 min**

````python
arr = [100,101,200,102,3,9]
s = set(arr)
max_length = 1
for i in arr:
    length = 1
    cur = i
    if cur-1 not in s:
        while cur+1 in s:
            length+=1
            cur+=1
    max_length=max(length,max_length)
    
max_length
````
