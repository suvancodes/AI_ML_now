## ⚠️ But here's the reality for a Tier-3 student

I don't want to give you the misleading impression that:

> **"I know Python + SQL + Power BI → I'll get ₹7 LPA."**

No.

For a fresher, ₹7 LPA+ is **above the normal market range**. Current estimates put many fresher DA offers around ₹3.5–6 LPA, although product companies, startups and stronger analytics firms can go substantially higher. ([GROWAI][4])

Your goal should therefore be:

**₹5–6 LPA = baseline target**
**₹7–8 LPA = strong target** 🎯
**₹9–12 LPA = aggressive target** 🔥

And because you're from a Tier-3 college, you need to compensate for the college-brand disadvantage with **off-campus applications + referrals + excellent SQL + strong projects + strong interview performance**.

### For your particular profile

You already have a fairly broad technical stack. Your next bottleneck isn't learning **20 more tools**.

Your bottleneck is becoming extremely good at:

1. **SQL** — joins, CTEs, windows, subqueries, optimization
2. **Python/Pandas**
3. **Power BI/Tableau**
4. **Statistics**
5. **Business case studies**
6. **Data Analyst interview questions**
7. **Communication**
8. **One genuinely impressive analytics project**
9. **Internship/real-world experience**
10. **DSA/basic coding for companies that test it**

If you nail those, **₹7 LPA+ becomes a legitimate target rather than wishful thinking.** 💪

One more thing: don't restrict yourself to job titles containing exactly **"Data Analyst."** Search for **Business Analyst, Product Analyst, BI Analyst, Analytics Analyst, Decision Scientist, Risk Analyst, Marketing Analyst, Operations Analyst** too. Some of these can pay better while giving you essentially the same analytical foundation.
