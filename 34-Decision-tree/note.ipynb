# 🌳 Decision Tree — Revision Notes

## 1. What is a Decision Tree?

A **Decision Tree** is a supervised machine learning algorithm that makes predictions by repeatedly splitting the dataset based on features.

It looks like a tree:

```text
                 Root
                /    \
             Split   Split
             /          \
          Split        Leaf
          /  \
       Leaf  Leaf
```

### Basic terminology

| Term              | Meaning                          |
| ----------------- | -------------------------------- |
| **Root Node**     | First/starting node              |
| **Internal Node** | A node where splitting happens   |
| **Branch**        | Result of a condition            |
| **Leaf Node**     | Final prediction                 |
| **Split**         | Dividing data based on a feature |

# 4. Pure vs Impure Split

### Pure node

All observations belong to the same class.

```text
Yes Yes Yes Yes
```

Impurity = **0**

### Impure node

Observations belong to different classes.

```text
Yes No Yes No
```

Impurity > **0**

### Goal of Decision Tree

> Create splits that make the resulting child nodes as **pure as possible**.

---

# 5. Entropy

Entropy measures the **impurity/uncertainty** of a node.

## Formula

For binary classification:

[
H(S)=-p_+\log_2(p_+)-p_-\log_2(p_-)
]

For multiple classes:

[
H(S)=-\sum_{i=1}^{n}p_i\log_2(p_i)
]

Where:

* (p_i) = proportion of class (i)
* (n) = number of classes

---

## Entropy Examples

### Pure node

Suppose:

```text
3 Yes
0 No
```

Then:

[
p_{Yes}=1,\quad p_{No}=0
]

Entropy:

[
H(S)=-1\log_2(1)-0\log_2(0)
]

[
\boxed{H(S)=0}
]

So:

> **Entropy = 0 → Pure node**

---

### Maximum impurity

For binary classification:

```text
50% Yes
50% No
```

Entropy:

[
\boxed{H(S)=1}
]

So:

```text
Entropy = 0     → Pure
Entropy = 1     → Maximum impurity (binary)
```

Your notes illustrate this relationship graphically.

---

# 6. Gini Impurity

Gini Impurity is another measure of impurity.

## Formula

[
Gini=1-\sum_{i=1}^{n}p_i^2
]

For binary classification:

[
Gini=1-(p_+^2+p_-^2)
]

### Example

Suppose:

```text
50% Yes
50% No
```

Then:

[
Gini=1-(0.5^2+0.5^2)
]

[
=1-(0.25+0.25)
]

[
=\boxed{0.5}
]

---

### Pure node

If:

```text
100% Yes
0% No
```

Then:

[
Gini=1-(1^2+0^2)
]

[
\boxed{Gini=0}
]

Therefore:

> **Gini = 0 → Pure node**

The calculations in your notes use exactly this idea.

| -------------- | -------------------- | -------------- |
| Measures       | Impurity             | Impurity       |
| Pure node      | 0                    | 0              |
| Binary maximum | 1                    | 0.5            |
| Formula        | (-\sum p_i\log_2p_i) | (1-\sum p_i^2) |

# 8. Information Gain ⭐

Information Gain tells us **which feature is better for splitting the data**.

### Formula

[
IG(S,F)
=======

## H(S)

\sum_v
\frac{|S_v|}{|S|}
H(S_v)
]

Where:

* (S) = original dataset
* (F) = feature
* (S_v) = subset produced by feature value (v)
* (H(S)) = entropy before splitting
* (H(S_v)) = entropy after splitting

### Intuition

```text
Before split
     ↓
High impurity
     ↓
Split using feature
     ↓
Lower impurity
     ↓
Good Information Gain
```

### Important:

> **Higher Information Gain → Better split**

Your notes compare candidate features using their Information Gain and select the feature with the larger gain.

# 13. How Regression Tree Predicts?

Suppose a leaf contains:

```text
40
42
```

The prediction is their average:

[
\frac{40+42}{2}
]

[
=\boxed{41}
]

So:

> **Regression Tree leaf prediction = Mean of target values in that leaf.**

# 14. Variance in Regression

For regression, your notes use **Variance** to evaluate the quality of a split.

## Formula

[
Variance=
\frac{1}{n}
\sum_{i=1}^{n}(y_i-\bar y)^2
]

Where:

* (y_i) = actual target value
* (\bar y) = mean of target values
* (n) = number of observations

---

# 15. Root Variance — Example

Suppose:

```text
[40, 42, 52, 60, 56]
```

Mean:

[
\bar y=50
]

Therefore:

[
Variance=
\frac{1}{5}
[
(40-50)^2+
(42-50)^2+
(52-50)^2+
(60-50)^2+
(56-50)^2
]
]

# [

\frac{1}{5}
[100+64+4+100+36]
]

[
\boxed{Variance=60.8}
]

This calculation is shown in your regression note.

---

# 16. Variance Reduction ⭐

For regression, we want to find a split that **reduces variance**.

## Formula

[
Variance\ Reduction
===================

## Var(root)

\sum_i w_i Var(child_i)
]

Where:

[
w_i=\frac{n_i}{n}
]

So:

```text
Parent Variance
       ↓
Compare with
weighted child variances
       ↓
Variance Reduction
```

### Goal:

> **Higher Variance Reduction → Better split**

---

\boxed{10.67}
]

# 19. Complete Decision Tree Flow

## Classification

```text
Dataset
   ↓
Calculate impurity
   ↓
Entropy / Gini
   ↓
Try different features
   ↓
Calculate Information Gain
   ↓
Choose best feature
   ↓
Split
   ↓
Repeat
   ↓
Leaf
   ↓
Class prediction
```

## Regression

```text
Dataset
   ↓
Calculate variance
   ↓
Try different features/thresholds
   ↓
Calculate Variance Reduction
   ↓
Choose best split
   ↓
Split
   ↓
Repeat
   ↓
Leaf
   ↓
Mean prediction
```
