# Gradient Boosting Regression — Revision Notes

## 1. What is Gradient Boosting?

**Gradient Boosting** is an ensemble learning algorithm.

It can be used for:

1. **Regression**
2. **Classification**

This note focuses on **Gradient Boosting Regression**.

The main idea is:

> Build models sequentially, where every new model tries to reduce the errors made by the previous model.

---

# 2. Basic Idea

Suppose we have a regression dataset:

| Experience | Degree  | Salary |
| ---------: | ------- | -----: |
|          2 | B.E     |    50K |
|          3 | Masters |    70K |
|          5 | Masters |    80K |
|          6 | PhD     |   100K |

The target variable is:

[
y = Salary
]

---

# 3. Step 1 — Create a Base Model

The first model is a very simple model.

For regression, the initial prediction can be the **mean of the target values**.

Given:

[
50,\ 70,\ 80,\ 100
]

The average is:

[
\frac{50+70+80+100}{4}
]

[
=\frac{300}{4}
]

[
\boxed{75}
]

Therefore, the initial/base prediction is:

[
\boxed{F_0(x)=75}
]

So initially every data point gets a prediction of:

```text
75
```

---

# 4. Step 2 — Calculate Residuals

Now calculate the difference between the actual value and the predicted value.

The residual is:

[
\boxed{
R_i=y_i-\hat{y}_i
}
]

where:

* (y_i) = actual value
* (\hat{y}_i) = predicted value
* (R_i) = residual/error

---

## Example

For the first sample:

Actual salary:

[
y=50K
]

Prediction:

[
\hat{y}=75K
]

Therefore:

[
R=50-75
]

[
\boxed{R=-25}
]

For the second sample:

[
R=70-75
]

[
\boxed{R=-5}
]

For the third sample:

[
R=80-75
]

[
\boxed{R=5}
]

For the fourth sample:

[
R=100-75
]

[
\boxed{R=25}
]

So the residuals are:

```text
Actual       Prediction       Residual
50K          75K              -25K
70K          75K               -5K
80K          75K               +5K
100K         75K              +25K
```

---

# 5. Step 3 — Train a Decision Tree on Residuals

This is the **key idea of Gradient Boosting Regression**.

Instead of directly predicting the original target (y), the next decision tree learns the **residuals/errors**.

So:

```text
Original Features
       +
Residuals
       ↓
Decision Tree
```

The tree learns:

[
(x_i,R_i)
]

instead of:

[
(x_i,y_i)
]

---

# 6. Why Train on Residuals?

Suppose our current prediction is:

[
75K
]

but the actual value is:

[
52K
]

Then:

[
R=52-75=-23K
]

The model is predicting **too high** by (23K).

The next tree learns this correction:

[
-23K
]

Therefore, the next prediction becomes:

[
75+(-23)
]

[
\boxed{52K}
]

This is the central intuition:

> **Each new tree learns how to correct the previous model's mistakes.**

---

# 7. Learning Rate

Gradient Boosting does **not** usually add the entire correction at once.

Instead, it uses a **learning rate**.

The notes use:

[
\boxed{\delta=0.1}
]

The learning rate controls how much of the new tree's prediction is added to the existing model.

---

# 8. Prediction Update

The prediction is updated using:

[
\boxed{
F_{m}(x)
========

F_{m-1}(x)
+
\delta h_m(x)
}
]

where:

* (F_{m-1}(x)) = previous model prediction
* (h_m(x)) = prediction of the new decision tree
* (\delta) = learning rate
* (F_m(x)) = updated prediction

---

## Example

Suppose:

[
F_0(x)=75
]

and the first tree predicts:

[
h_1(x)=-23
]

with:

[
\delta=0.1
]

Then:

[
F_1(x)
======

75+(0.1)(-23)
]

[
=75-2.3
]

[
\boxed{F_1(x)=72.7}
]

So instead of immediately changing:

[
75\rightarrow52
]

Gradient Boosting makes a smaller correction:

[
75\rightarrow72.7
]

This helps the model learn gradually.

---

# 9. Repeat the Process

The process continues sequentially.

```text
Base Model
    ↓
Calculate Residuals
    ↓
Train Decision Tree on Residuals
    ↓
Multiply Tree Prediction by Learning Rate
    ↓
Update Prediction
    ↓
Calculate New Residuals
    ↓
Train Next Tree
    ↓
Repeat
```

Each new tree attempts to correct the remaining errors.

---

# 10. Mathematical Form of Gradient Boosting

Suppose we have (n) trees:

[
h_0(x),h_1(x),h_2(x),\ldots,h_n(x)
]

The final model can be written as:

[
\boxed{
F(x)
====

\alpha_0h_0(x)
+
\alpha_1h_1(x)
+
\alpha_2h_2(x)
+\cdots+
\alpha_nh_n(x)
}
]

In the notes, the learning rate is:

[
\delta=0.1
]

So the model can be represented as:

[
\boxed{
F(x)
====

\sum_{i=0}^{n}
\delta h_i(x)
}
]

with the initial/base model included appropriately.

---

# 11. Complete Example

Suppose the initial prediction is:

[
F_0(x)=75
]

The first tree predicts:

[
h_1(x)=-23
]

Learning rate:

[
\delta=0.1
]

Then:

[
F_1(x)
======

75+0.1(-23)
]

[
=72.7
]

Suppose the next tree predicts another correction:

[
h_2(x)=-3
]

Then:

[
F_2(x)
======

72.7+0.1(-3)
]

[
=72.4
]

The process continues:

[
F_3(x)
======

F_2(x)+0.1h_3(x)
]

and so on.

---

# 12. Gradient Boosting Intuition

Think of Gradient Boosting like this:

```text
Initial Prediction
       ↓
     75K
       ↓
"Where am I wrong?"
       ↓
Residuals
       ↓
Tree 1 learns corrections
       ↓
Small correction
       ↓
Better Prediction
       ↓
"Where am I still wrong?"
       ↓
Residuals
       ↓
Tree 2 learns corrections
       ↓
Small correction
       ↓
Even Better Prediction
       ↓
...
```

### Core idea

> **Gradient Boosting builds an additive model where each new weak learner tries to correct the errors of the existing model.**

---

# 13. Gradient Boosting vs AdaBoost

| Feature        | AdaBoost                       | Gradient Boosting                               |
| -------------- | ------------------------------ | ----------------------------------------------- |
| Main idea      | Focus on misclassified samples | Learn previous model's errors                   |
| Weighting      | Changes sample weights         | Fits next model to residuals/negative gradients |
| Training       | Sequential                     | Sequential                                      |
| Base learners  | Often decision stumps          | Usually shallow decision trees                  |
| Learning rate  | Can be used                    | Very important                                  |
| Main intuition | Focus on difficult samples     | Correct remaining errors                        |

### Easy way to remember

```text
AdaBoost
    ↓
Increase importance of mistakes

Gradient Boosting
    ↓
Learn the mistakes/residuals
```

---

# 14. Important Terms

### Base Model

The initial/simple model.

For regression:

[
\boxed{
F_0(x)=\text{Mean}(y)
}
]

---

### Residual

Difference between actual and predicted value:

[
\boxed{
R_i=y_i-\hat{y}_i
}
]

---

### Weak Learner

A small/shallow decision tree that learns the residuals.

---

### Learning Rate

Controls how strongly each new tree affects the final prediction.

[
\boxed{\delta=0.1}
]

in the example.

---

### Final Model

The final prediction is the sum of the contributions from the base model and subsequent trees.

---

# 15. Must-Know Formulas

### Base Prediction

[
\boxed{
F_0(x)=\frac{1}{N}\sum_{i=1}^{N}y_i
}
]

---

### Residual

[
\boxed{
R_i=y_i-F_{m-1}(x_i)
}
]

---

### Prediction Update

[
\boxed{
F_m(x)
======

F_{m-1}(x)
+
\delta h_m(x)
}
]

---

### Final Model

[
\boxed{
F(x)
====

F_0(x)
+
\delta h_1(x)
+
\delta h_2(x)
+\cdots+
\delta h_M(x)
}
]

or:

[
\boxed{
F(x)
====

F_0(x)
+
\delta\sum_{m=1}^{M}h_m(x)
}
]

---

# 16. One-Minute Revision 🚀

Remember these **6 steps**:

1. **Calculate the mean of the target** → Base prediction.
2. **Calculate residuals**:

[
R=y-\hat y
]

3. **Train a decision tree on residuals.**
4. Multiply tree prediction by the **learning rate**.
5. Add the correction to the previous prediction:

[
F_m=F_{m-1}+\delta h_m
]

6. **Repeat** until the ensemble becomes strong.

### One-line intuition

> **Gradient Boosting starts with a simple prediction and repeatedly adds small decision-tree corrections to reduce the remaining errors.** 🎯
