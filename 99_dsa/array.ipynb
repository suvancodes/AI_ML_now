### chack array sorted of not

```python
arr = [1,2,3,94,5,6]
o = 1
s = True
for z in range(len(arr)-1):
    if arr[z]>arr[o]:
        s=False
    o = o+1
    
s
```

### reverce a array

```python
arr = [1,2,3,94,5,6]
st = 0
end = len(arr)-1
while st<=end:
    arr[st],arr[end] = arr[end],arr[st]
    st = st+1
    end = end-1
    
arr
```

### remove duplicate

```python
arr = [1, 2, 2, 3, 3,89,99,99]
n=[]
u=0
for i in range(len(arr)):
    if arr[u] == arr[i]:
        continue
    n.append(arr[u])
    u=i
        
        
n
```

### Move All Zeros to End

```python
arr = [0, 2, 0, 3, 0,0,99,99]
n = len(arr)-1
for i in range(len(arr)-1,-1,-1):
    if arr[i] == 0:
        arr[n],arr[i] = arr[i],arr[n]
        n-=1
arr
    
```
